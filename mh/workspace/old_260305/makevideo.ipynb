```python
import cv2
import os

def images_to_video(folder_path, output_name, fps):
    # 1. 폴더 내 이미지 파일 목록 가져오기 (이미지 확장자 필터링)
    images = [img for img in os.listdir(folder_path) 
              if img.endswith((".png", ".jpg", ".jpeg"))]
    
    # 2. 파일 이름 순서대로 정렬 (숫자 순서 등을 위해 중요!)
    images.sort()

    if not images:
        print("폴더에 이미지 파일이 없습니다.")
        return

    # 3. 첫 번째 이미지를 읽어 영상의 가로, 세로 크기 설정
    frame = cv2.imread(os.path.join(folder_path, images[0]))
    height, width, layers = frame.shape

    # 4. 비디오 라이터 설정 (MP4 코덱 사용)
    fourcc = cv2.VideoWriter_fourcc(*'mp4v') 
    video = cv2.VideoWriter(output_name, fourcc, fps, (width, height))

    print(f"{len(images)}개의 이미지로 영상 제작을 시작합니다...")

    # 5. 이미지를 하나씩 읽어 영상에 추가
    for image in images:
        img_path = os.path.join(folder_path, image)
        frame = cv2.imread(img_path)
        
        # 이미지 크기가 첫 번째 이미지와 다를 경우를 대비해 리사이즈 (선택 사항)
        # frame = cv2.resize(frame, (width, height))
        
        video.write(frame)

    # 6. 작업 완료 후 해제
    video.release()
    print(f"영상 제작 완료: {output_name}")

# --- 설정 값 ---
folder_name = 'a'            # 이미지 폴더 경로
output_video = 'result.mp4'  # 결과 영상 파일명
frames_per_second = 24       # 초당 프레임 수 (FPS)

images_to_video(folder_name, output_video, frames_per_second)
```

```python
import cv2
import os

def create_car_video():
    folder_path = 'uav'
    output_name = 'uav.mp4'
    
    # 1. 파일 목록 생성 및 정렬 (snapshot_0001 ~ 0019)
    # 특정 패턴의 파일만 가져오기 위해 리스트 컴프리헨션 사용
    images = [img for img in os.listdir(folder_path) 
              if img.startswith("snapshot_") and img.endswith(".png")]
    images.sort() # 이름순 정렬

    if not images:
        print(f"'{folder_path}' 폴더에 조건에 맞는 이미지가 없습니다.")
        return

    # 2. 영상 속성 설정
    # 19장을 4초 동안 보여주기 위한 FPS 계산
    fps = 19 / 4  # 4.75
    
    # 첫 번째 이미지로 해상도 읽기
    first_frame = cv2.imread(os.path.join(folder_path, images[0]))
    height, width, layers = first_frame.shape

    # 비디오 라이터 초기화 (MP4 코덱)
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    video = cv2.VideoWriter(output_name, fourcc, fps, (width, height))

    print(f"영상 제작 시작: {output_name} (FPS: {fps})")

    # 3. 이미지 프레임 추가
    target_w, target_h = 1920, 1080   # 16:9
    video = cv2.VideoWriter(output_name, fourcc, fps, (target_w, target_h))

    for image in images:
        frame = cv2.imread(os.path.join(folder_path, image))
        if frame is None:
            continue
        frame = cv2.resize(frame, (target_w, target_h))  # 강제 변환
        video.write(frame)

    # 4. 종료
    video.release()
    print("--- 영상 제작이 완료되었습니다 ---")

if __name__ == "__main__":
    create_car_video()
```
