```python
import cv2
import os
import glob

def make_video(source_dir, output_path, fps=15):
    # 1. 파일 경로 설정 (frame_0000.png 패턴)
    # glob을 사용하면 패턴 매칭이 쉽습니다.
    img_array = []
    
    # 0000 ~ 0144 순서대로 가져오기 위해 정렬
    # 이름이 0000 형태라 일반 sort()만 해도 순서가 잘 맞습니다.
    files = sorted(glob.glob(os.path.join(source_dir, 'frame_*.png')))
    
    if not files:
        print("이미지 파일을 찾을 수 없습니다.")
        return

    # 2. 첫 번째 이미지로 비디오 사이즈 결정
    img = cv2.imread(files[0])
    height, width, layers = img.shape
    size = (width, height)
    
    # 3. 비디오 생성기 설정
    out = cv2.VideoWriter(output_path, cv2.VideoWriter_fourcc(*'mp4v'), fps, size)
    
    print(f"총 {len(files)}장의 이미지를 변환합니다...")
    
    # 4. 이미지 쓰기
    for filename in files:
        img = cv2.imread(filename)
        out.write(img)
        
    out.release()
    print(f"완료! 저장된 파일: {output_path}")

# --- 실행 부분 ---
if __name__ == "__main__":
    # 현재 사용자분의 경로
    current_path = '/data1/mh/sionna/ext/sionna-rt/tutorials/sim_frames'
    output_file = os.path.join(current_path, 'sionna_simulation.mp4')
    
    # FPS 15 추천 (너무 빠르면 시뮬레이션 내용이 안 보임)
    make_video(current_path, output_file, fps=15)
```
